# Mathematics

## The expressions from Zhu et al. Box 1 and Supplementary File

The paper assumes the **same dissociation constant** $K_d$ for AA, BB and AB. After non-dimensionalization, its equilibrium relationships are

$$
[A]^2=K_d[A_2],\qquad [B]^2=K_d[B_2],\qquad 2[A][B]=K_d[AB].
$$

Mass conservation gives

$$
[A_{\mathrm{tot}}]=[A]+2[A_2]+[AB],\qquad
[B_{\mathrm{tot}}]=[B]+2[B_2]+[AB].
$$

**Substitute the mass-action equations:**

$$
\begin{aligned}
[A_{\mathrm{tot}}]&=[A]+\frac{2[A]^2}{K_d}+\frac{2[A][B]}{K_d}
=[A]\left(1+\frac{2([A]+[B])}{K_d}\right),\\
[B_{\mathrm{tot}}]&=[B]+\frac{2[B]^2}{K_d}+\frac{2[A][B]}{K_d}
=[B]\left(1+\frac{2([A]+[B])}{K_d}\right).
\end{aligned}
$$

Because **the factor in parentheses is identical in both equations**, the paper can solve for both homodimer concentrations explicitly:

$$
\boxed{
[A_2]=\frac{2[A_{\mathrm{tot}}]^2}
{K_d+4([A_{\mathrm{tot}}]+[B_{\mathrm{tot}}])+\sqrt{K_d^2+8K_d([A_{\mathrm{tot}}]+[B_{\mathrm{tot}}])}}
}
$$

$$
\boxed{
[B_2]=\frac{2[B_{\mathrm{tot}}]^2}
{K_d+4([A_{\mathrm{tot}}]+[B_{\mathrm{tot}}])+\sqrt{K_d^2+8K_d([A_{\mathrm{tot}}]+[B_{\mathrm{tot}}])}}
}
$$

Then we insert these into **paper's dimensionless ODE structure** which is defined below. This would give us everything given in terms of total A and B:

$$
\boxed{\begin{aligned}
\frac{d[A_{\mathrm{tot}}]}{dt}&=\alpha+\beta\frac{[A_2]^n}{1+[A_2]^n}-[A_{\mathrm{tot}}],\\[5pt]
\frac{d[B_{\mathrm{tot}}]}{dt}&=\alpha+\beta\frac{[B_2]^n}{1+[B_2]^n}-[B_{\mathrm{tot}}].
\end{aligned}}
$$

## Modified Model with Different Dimerization Affinities

We relax one assumption by allowing the three dimerization reactions to have independent dissociation constants:

| Original model | Modified model |
|---|---|
| $[A]^2=K_d[A_2]$ | $[A]^2=K_{AA}[A_2]$ |
| $[B]^2=K_d[B_2]$ | $[B]^2=K_{BB}[B_2]$ |
| $2[A][B]=K_d[AB]$ | $2[A][B]=K_{AB}[AB]$ |

The factor of 2 in the heterodimer equilibrium equation is retained from the authors' convention.

**Rearranging the modified equilibrium equations gives:**

$$
\boxed{[A_2]=\frac{[A]^2}{K_{AA}}}
$$

$$
\boxed{[B_2]=\frac{[B]^2}{K_{BB}}}
$$

$$
\boxed{[AB]=\frac{2[A][B]}{K_{AB}}}
$$

Our objective is to express the concentrations of the activating homodimers, $[A_2]$ and $[B_2]$, in terms of the total transcription factor concentrations, $[A_{\mathrm{tot}}]$ and $[B_{\mathrm{tot}}]$, so that they can be substituted into the original ODEs.

## Applying conservation of mass


The conservation equations remain unchanged:

$$
[A_{\mathrm{tot}}]=[A]+2[A_2]+[AB]
$$

$$
[B_{\mathrm{tot}}]=[B]+2[B_2]+[AB]
$$

Each transcription factor can exist as a free monomer, as part of a homodimer, or as part of a heterodimer.

Substituting the modified mass-action equations gives:

$$
\boxed{
[A_{\mathrm{tot}}]=[A]+\frac{2[A]^2}{K_{AA}}+\frac{2[A][B]}{K_{AB}}
}
$$

$$
\boxed{
[B_{\mathrm{tot}}]=[B]+\frac{2[B]^2}{K_{BB}}+\frac{2[A][B]}{K_{AB}}
}
$$

For comparison:

| Original model | Modified model |
|---|---|
| $[A_{\mathrm{tot}}]=[A]+\frac{2[A]^2}{K_d}+\frac{2[A][B]}{K_d}$ | $[A_{\mathrm{tot}}]=[A]+\frac{2[A]^2}{K_{AA}}+\frac{2[A][B]}{K_{AB}}$ |
| $[B_{\mathrm{tot}}]=[B]+\frac{2[B]^2}{K_d}+\frac{2[A][B]}{K_d}$ | $[B_{\mathrm{tot}}]=[B]+\frac{2[B]^2}{K_{BB}}+\frac{2[A][B]}{K_{AB}}$ |


## Why can we not use the original analytical solution?

In the original model, both conservation equations can be factorized using the same expression:

$$
[A_{\mathrm{tot}}]
=
[A]\left(1+\frac{2([A]+[B])}{K_d}\right)
$$

$$
[B_{\mathrm{tot}}]
=
[B]\left(1+\frac{2([A]+[B])}{K_d}\right)
$$

Adding these equations produces:

$$
[A_{\mathrm{tot}}]+[B_{\mathrm{tot}}]
=
([A]+[B])\left(1+\frac{2([A]+[B])}{K_d}\right)
$$

This can be rearranged into a quadratic equation:

$$
2([A]+[B])^2
+
K_d([A]+[B])
-
K_d([A_{\mathrm{tot}}]+[B_{\mathrm{tot}}])
=0
$$

Solving this quadratic and substituting the result into the homodimer equilibrium equations gives the authors' original expressions:

$$
\boxed{
[A_2]=
\frac{2[A_{\mathrm{tot}}]^2}{
K_d+4([A_{\mathrm{tot}}]+[B_{\mathrm{tot}}])
+\sqrt{K_d^2+8K_d([A_{\mathrm{tot}}]+[B_{\mathrm{tot}}])}
}
}
$$

$$
\boxed{
[B_2]=
\frac{2[B_{\mathrm{tot}}]^2}{
K_d+4([A_{\mathrm{tot}}]+[B_{\mathrm{tot}}])
+\sqrt{K_d^2+8K_d([A_{\mathrm{tot}}]+[B_{\mathrm{tot}}])}
}
}
$$

In our modified model, however, factorization gives:

$$
[A_{\mathrm{tot}}]
=
[A]\left(
1+\frac{2[A]}{K_{AA}}+\frac{2[B]}{K_{AB}}
\right)
$$

$$
[B_{\mathrm{tot}}]
=
[B]\left(
1+\frac{2[B]}{K_{BB}}+\frac{2[A]}{K_{AB}}
\right)
$$

The expressions multiplying $[A]$ and $[B]$ are now different. So no more quadratic formula.
Therefore, **the original analytical expressions for $[A_2]$ and $[B_2]$ cannot be reused when the dissociation constants differ**. Instead, we solve the modified dimerization equilibrium numerically.

We can use the heterodimer concentration $[AB]$, which is already present in the original model, as an intermediate unknown. (This idea comes from OpenAI ChatGPT-6)

### Expressing free [A] in terms of total A and [AB] and free [B] in terms of total B and [AB]



$$
[A_{\mathrm{tot}}]
=
[A]+\frac{2[A]^2}{K_{AA}}+[AB]
$$

Move the heterodimer concentration to the left-hand side:

$$
[A_{\mathrm{tot}}]-[AB]
=
[A]+\frac{2[A]^2}{K_{AA}}
$$

Multiply both sides by $K_{AA}$:

$$
K_{AA}([A_{\mathrm{tot}}]-[AB])
=
K_{AA}[A]+2[A]^2
$$

Rearrange:

$$
\boxed{
2[A]^2+K_{AA}[A]-K_{AA}([A_{\mathrm{tot}}]-[AB])=0
}
$$

This is a quadratic equation in the free monomer concentration $[A]$.

$$
[A]=
\frac{
-K_{AA}\pm
\sqrt{
K_{AA}^2+8K_{AA}([A_{\mathrm{tot}}]-[AB])
}
}{4}
$$

Since concentrations must be nonnegative, we select the positive root:

$$
\boxed{
[A]=
\frac{
-K_{AA}+
\sqrt{
K_{AA}^2+8K_{AA}([A_{\mathrm{tot}}]-[AB])
}
}{4}
}
$$

For numerical calculations, we can rearrange this expression by rationalizing the numerator. The second form is used in the code because it avoids subtracting nearly equal numbers when the free monomer concentration is small. (The idea comes from OpenAI's ChatGPT-6 ):

$$
\boxed{
[A]=
\frac{
2([A_{\mathrm{tot}}]-[AB])
}{
1+\sqrt{1+\frac{8([A_{\mathrm{tot}}]-[AB])}{K_{AA}}}
}
}
$$




We can now calculate $[A]$ if we know $[A_{\mathrm{tot}}]$, $K_{AA}$, and $[AB]$. We follow the same procedure for transcription factor B.


$$
[B_{\mathrm{tot}}]=[B]+2[B_2]+[AB]
$$

$$
[B_2]=\frac{[B]^2}{K_{BB}}
$$

$$
[B_{\mathrm{tot}}]
=
[B]+\frac{2[B]^2}{K_{BB}}+[AB]
$$

$$
[B_{\mathrm{tot}}]-[AB]
=
[B]+\frac{2[B]^2}{K_{BB}}
$$

Multiply by $K_{BB}$:

$$
2[B]^2+K_{BB}[B]
-K_{BB}([B_{\mathrm{tot}}]-[AB])=0
$$

Applying the quadratic formula and selecting the non-negative root:

$$
[B]=
\frac{
-K_{BB}+
\sqrt{
K_{BB}^2+8K_{BB}([B_{\mathrm{tot}}]-[AB])
}
}{4}
$$

$$
\boxed{
[B]=
\frac{
2([B_{\mathrm{tot}}]-[AB])
}{
1+\sqrt{1+\frac{8([B_{\mathrm{tot}}]-[AB])}{K_{BB}}}
}
}
$$

We now have expressions for both free monomer concentrations in terms of the total concentrations and the heterodimer concentration. The remaining unknown is $[AB]$.

### Determining the heterodimer concentration



The modified heterodimerization equilibrium is:

$$
2[A][B]=K_{AB}[AB]
$$

Therefore:

$$
\boxed{
[AB]=\frac{2[A][B]}{K_{AB}}
}
$$

We have already expressed $[A]$ and $[B]$ in terms of $[AB]$, so we can substitute those expressions into this equilibrium equation. This gives:

$$
\boxed{
[AB]=
\frac{2}{K_{AB}}
\left(
\frac{
2([A_{\mathrm{tot}}]-[AB])
}{
1+\sqrt{1+\frac{8([A_{\mathrm{tot}}]-[AB])}{K_{AA}}}
}
\right)
\left(
\frac{
2([B_{\mathrm{tot}}]-[AB])
}{
1+\sqrt{1+\frac{8([B_{\mathrm{tot}}]-[AB])}{K_{BB}}}
}
\right)
}
$$

**This equation contains only one unknown: $[AB]$.** All other quantities are known for a chosen pair of total concentrations and dissociation constants. Therefore, rather than solving two coupled nonlinear equations simultaneously for $[A]$ and $[B]$, we can solve a single nonlinear equation for $[AB]$.

### Numerical solution using bisection



The heterodimer concentration must satisfy:

$$
\boxed{
0\leq[AB]\leq
\min([A_{\mathrm{tot}}],[B_{\mathrm{tot}}])
}
$$

For any selected value of $[AB]$, we can calculate the corresponding free monomer concentrations using the expressions derived above. We then check whether:

$$
\boxed{
[AB]-\frac{2[A][B]}{K_{AB}}=0
}
$$

The numerical implementation uses **bisection** to find the heterodimer concentration satisfying this equation (OpenAI's ChatGPT-6). Bisection repeatedly divides the physically allowed interval into two halves and retains the half containing the solution. 

The numerical calculation therefore follows these steps:

1. Choose total concentrations $[A_{\mathrm{tot}}]$ and $[B_{\mathrm{tot}}]$.
2. Use the bisection method to determine $[AB]$.
3. Substitute the calculated $[AB]$ into the expressions for free $[A]$ and $[B]$.
4. Calculate the homodimer concentrations from the modified mass-action equations.


### Obtaining [A₂] and [B₂] in terms of total concentrations



Once $[AB]$ has been determined, we calculate the free monomer concentrations:

$$
[A]=
\frac{
2([A_{\mathrm{tot}}]-[AB])
}{
1+\sqrt{1+\frac{8([A_{\mathrm{tot}}]-[AB])}{K_{AA}}}
}
$$

$$
[B]=
\frac{
2([B_{\mathrm{tot}}]-[AB])
}{
1+\sqrt{1+\frac{8([B_{\mathrm{tot}}]-[AB])}{K_{BB}}}
}
$$

From mass-action equilibrium:

$$
[A_2]=\frac{[A]^2}{K_{AA}}
$$

Substituting the expression for $[A]$:

$$
[A_2]=
\frac{1}{K_{AA}}
\left(
\frac{
2([A_{\mathrm{tot}}]-[AB])
}{
1+\sqrt{1+\frac{8([A_{\mathrm{tot}}]-[AB])}{K_{AA}}}
}
\right)^2
$$

$$
\boxed{
[A_2]=
\frac{
4([A_{\mathrm{tot}}]-[AB])^2
}{
K_{AA}
\left(
1+\sqrt{1+\frac{8([A_{\mathrm{tot}}]-[AB])}{K_{AA}}}
\right)^2
}
}
$$

Similarly:

$$
[B_2]=
\frac{1}{K_{BB}}
\left(
\frac{
2([B_{\mathrm{tot}}]-[AB])
}{
1+\sqrt{1+\frac{8([B_{\mathrm{tot}}]-[AB])}{K_{BB}}}
}
\right)^2
$$

$$
\boxed{
[B_2]=
\frac{
4([B_{\mathrm{tot}}]-[AB])^2
}{
K_{BB}
\left(
1+\sqrt{1+\frac{8([B_{\mathrm{tot}}]-[AB])}{K_{BB}}}
\right)^2
}
}
$$

These are the modified expressions for the activating homodimer concentrations.

Unlike the original analytical expressions, they still contain $[AB]$. However, $[AB]$ is **not an independent dynamic variable**. Consequently, both homodimer concentrations are completely determined by the total concentrations and model parameters.

### Substituting the modified homodimer concentrations into the ODEs


We are rescaling new constants by $K_M$ and time by the inverse protein-removal rate as it was done by the paper (Check Supplementary File). So we have new dimensionless quantities. 

$$
K_{AA}\leftarrow\frac{K_{AA}}{K_M},
\qquad
K_{BB}\leftarrow\frac{K_{BB}}{K_M},
\qquad
K_{AB}\leftarrow\frac{K_{AB}}{K_M}
$$

The authors' dimensionless ODEs are:

$$
\frac{d[A_{\mathrm{tot}}]}{dt}
=
\alpha+
\beta\frac{[A_2]^n}{1+[A_2]^n}
-[A_{\mathrm{tot}}]
$$

$$
\frac{d[B_{\mathrm{tot}}]}{dt}
=
\alpha+
\beta\frac{[B_2]^n}{1+[B_2]^n}
-[B_{\mathrm{tot}}]
$$

The structure of these equations remains unchanged because we have not modified the assumptions regarding basal production, self-activation, or protein degradation. Using:

$$
[A_2]=\frac{[A]^2}{K_{AA}},
\qquad
[B_2]=\frac{[B]^2}{K_{BB}}
$$

We obtain:

$$
\boxed{
\frac{d[A_{\mathrm{tot}}]}{dt}
=
\alpha+
\beta
\frac{
\left(\frac{[A]^2}{K_{AA}}\right)^n
}{
1+\left(\frac{[A]^2}{K_{AA}}\right)^n
}
-[A_{\mathrm{tot}}]
}
$$

$$
\boxed{
\frac{d[B_{\mathrm{tot}}]}{dt}
=
\alpha+
\beta
\frac{
\left(\frac{[B]^2}{K_{BB}}\right)^n
}{
1+\left(\frac{[B]^2}{K_{BB}}\right)^n
}
-[B_{\mathrm{tot}}]
}
$$

In these equations, $[A]$ and $[B]$ are calculated from the total concentration Thus, the modified ODEs still describe a system with only two dynamic variables:

$$
[A_{\mathrm{tot}}],\qquad[B_{\mathrm{tot}}]
$$

### Python variables and their mathematical meaning



#### 1. Transcription factor concentrations

| Python variable | Mathematical notation | Meaning |
|---|---|---|
| `Atot` | $[A_{\mathrm{tot}}]$ | Total concentration of transcription factor A, including free A, AA homodimers, and AB heterodimers. |
| `Btot` | $[B_{\mathrm{tot}}]$ | Total concentration of transcription factor B, including free B, BB homodimers, and AB heterodimers. |
| `A` | $[A]$ | Concentration of free A monomers that are not part of a dimer. |
| `B` | $[B]$ | Concentration of free B monomers that are not part of a dimer. |
| `A2` | $[A_2]$ | Concentration of AA homodimers, which activate the production of A. |
| `B2` | $[B_2]$ | Concentration of BB homodimers, which activate the production of B. |
| `AB` | $[AB]$ | Concentration of inactive AB heterodimers, which sequester A and B from forming activating homodimers. |

#### 2. Model parameters

| Python variable | Mathematical notation | Meaning |
|---|---|---|
| `p.alpha` | $\alpha$ | Basal production rate of A, even without homodimer activation. |
| `p.beta` | $\beta$ | Maximum additional production rate resulting from homodimer activation. |
| `p.n` | $n$ | Hill coefficient controlling how sharply transcriptional activation increases with homodimer concentration. |
| `p.Kd` | $K_d$ | Original dissociation constant, assumed identical for AA, BB, and AB. |
| `p.KAA` | $K_{AA}$ | Dissociation constant for AA homodimerization in modified model. |
| `p.KBB` | $K_{BB}$ | Dissociation constant for BB homodimerization in modified model. |
| `p.KAB` | $K_{AB}$ | Dissociation constant for AB heterodimerization in modified model. |
| `p.r` | $r$ | Multiplier for B's basal production relative to A. |
| `p.m` | $m$ | Multiplier for B's activation-dependent production relative to A. |
| `p.kappa` | $\kappa$ | Relative activation threshold of B compared with A. |
| `p.gamma` | $\gamma$ | Relative degradation rate of B compared with A. |

#### 3. Variables used in numerical calculations

| Python variable | Mathematical meaning | Purpose |
|---|---|---|
| `y` in `df_dt(y, t, p)` | $([A_{\mathrm{tot}}],[B_{\mathrm{tot}}])$ | A pair containing both total concentrations, passed to the ODE function. |
| `t` | $t$ | Time variable used by the ODE solver. |
| `p` | Model parameter object | Stores the parameter values used in a particular simulation. |
| `dA` | $d[A_{\mathrm{tot}}]/dt$ | Instantaneous rate of change of total A. |
| `dB` | $d[B_{\mathrm{tot}}]/dt$ | Instantaneous rate of change of total B. |
| `lo` | Lower bound for $[AB]$ | Smallest possible heterodimer concentration remaining during bisection. |
| `hi` | Upper bound for $[AB]$ | Largest possible heterodimer concentration remaining during bisection. |
| `error` | $[AB]-2[A][B]/K_{AB}$ | Checks whether the guessed heterodimer concentration satisfies the equilibrium equation. |
| `steps` in `modified_dimers()` | Number of bisection iterations | Determines how many times the numerical search interval is divided in half. |
| `x0`, `y0` | $[A_{\mathrm{tot}}]$, $[B_{\mathrm{tot}}]$ at a particular point | Coordinates where the original Jacobian and stability are evaluated. |
| `mass_matrix` | Derivatives of the conservation equations | Used to calculate how changes in total concentrations affect free monomer concentrations in the modified Jacobian. |

#### 4. Variables controlling the plots

| Python variable | Meaning |
|---|---|
| `REGIMES` | Stores the original paper's parameter combinations for monostability, bistability, tristability, and quadrastability. |
| `AFFINITY_CASES` | Defines our modified experiments by specifying $K_{AA}$, $K_{BB}$, and $K_{AB}$ independently. |
| `RELATIVE_TO_ORIGINAL_KD` | Determines whether the values in `AFFINITY_CASES` are multiples of the original $K_d$ or absolute dimensionless values. |
| `Amax`, `Bmax` | Maximum total A and B concentrations displayed on the phase portraits. |
| `roots` | Fixed points found by the numerical solver. |
| `stable` | Fixed points classified as stable using the Jacobian eigenvalues. |
| `ax` | Matplotlib axis on which a phase portrait is drawn. |

```python
def plot_flow_field(ax, duv_dt, u_range, v_range, p, density=1000, sep_fun=None, figscale=1):
    """
    Plots the flow field with line thickness proportional to speed.
    Plots the nullclines
    Plots the stable and unstable fixed points
    Plots the separatrices (optional)
    """
    # for calculating separatrix in the end
    t = np.linspace(0,50,51) # `np.linspace(start, stop, count)` makes equally spaced time points.
    
    ### First calculate and plot the flow field (phase portrait)
    # set up u,v space
    u = np.linspace(u_range[0], u_range[1], density) # Choose the A-total values drawn along the horizontal axis.
    v = np.linspace(v_range[0], v_range[1], density) # Choose the B-total values drawn along the vertical axis.

    u_vel = np.zeros((len(u), len(v)))
    v_vel = np.zeros((len(u), len(v)))
    uu, vv = np.meshgrid(u, v) # `meshgrid` creates every pair of A and B values: a 2-D concentration grid.

    # calculate velocity field
    # Outputs are d[A_total]/dt and d[B_total]/dt: the arrows' directions.
    u_vel, v_vel = duv_dt([uu, vv], t, p) # Run the ODE function on EVERY grid point at once.
    
    # compute speed
    speed = np.sqrt(u_vel**2 + v_vel**2) #Arrow speed = sqrt((dA/dt)^2 + (dB/dt)^2).

    # make linewidths proportional to speed, with minimal line width of 0.5 and max of 5
    lw = 0.5 + 4.5 * speed / speed.max() # Faster changes get thicker flow lines; slow changes get thinner lines.
    
    # make stream plot
    s1 = ax.streamplot(uu, vv, u_vel, v_vel, linewidth=lw*figscale, arrowsize=1.5*figscale,
                   density=1, color='grey', zorder=1) # `ax.streamplot` draws gray curves showing where the ODE carries the system.
    
    ### Then calculate and plot the nullclines
    # plot nullclines
    c1 = ax.contour(uu, vv, u_vel, 0, colors=sns.color_palette()[1], linewidths=3*figscale, zorder=2)
    c1_label, _ = c1.legend_elements()
    c2 = ax.contour(uu, vv, v_vel, 0, colors=sns.color_palette()[3], linewidths=3*figscale, zorder=2)
    c2_label, _ = c2.legend_elements()
    
    # get nullclines
    if len(c1.allsegs[1])==1:
        Anull = c1.allsegs[1][0]
    else:
        # concatenate all A nullcline segments
        Anull = c1.allsegs[1][0]
        for i in range(len(c1.allsegs[1])-1):
            Anull = np.concatenate((Anull, c1.allsegs[1][i+1]))
        Anull = Anull[np.argsort(Anull[:,0])]
    
    if len(c2.allsegs[1])==1:
        Bnull = c2.allsegs[1][0]
    else:
        # concatenate all B nullcline segments
        Bnull = c2.allsegs[1][0]
        for i in range(len(c2.allsegs[1])-1):
            Bnull = np.concatenate((Bnull, c2.allsegs[1][i+1]))
        Bnull = Bnull[np.argsort(Bnull[:,1])]

    ### Calculate and plot the fixed points
    
    # find fixed points 
    # The two nullclines cross only where BOTH rates are zero:
    x, y = intersection(Anull[:,0], Anull[:,1], Bnull[:,0], Bnull[:,1])
    stable_fp = []
    unstable_fp = []
    # for each fixed point, use the surrounding 4 points to determine the stability
    for i in range(len(x)):
        if stability(x[i], y[i], p)==1:
            stable_fp.append([x[i], y[i]])
        else:
            unstable_fp.append([x[i], y[i]])

    # plot stable fixed points
    c3 = ax.scatter(np.array(stable_fp)[:,0], np.array(stable_fp)[:,1], zorder=5,
                     c='black', s=600*figscale, edgecolors='black', linewidth=3*figscale) # `ax.scatter` draws stable fixed points as filled black circles.
    sep = []
    # plot unstable fixed points
    if len(unstable_fp)>0:
        c4 = ax.scatter(np.array(unstable_fp)[:,0], np.array(unstable_fp)[:,1], zorder=4,
                     c='white', s=600*figscale, edgecolors='black', linewidth=3*figscale)
        
        ### Calculate and plot the separatrices
        if sep_fun!=None:
            # calculate the smallest stepsize between adjacent phase portrait points
            stepsize = u[1] - u[0]
            # calculate and plot the seperatrices
            for i in range(len(unstable_fp)):
                sep.append(sep_fun(unstable_fp[i][0], unstable_fp[i][1],
                                   duv_dt, p, stepsize, u_range[1], v_range[1]))
            
            for i in range(len(sep)):   
                c5 = ax.plot(sep[i][:,0], sep[i][:,1], '--', linewidth=3*figscale, color=sns.color_palette()[2], zorder=3)
            
            legend = ax.legend((c1_label[0], c2_label[0], c3, c4, c5[0]),
                               ('Nullcline dA/dt = 0','Nullcline dB/dt = 0', 'Stable fixed point', 'Unstable fixed point', 'Separatrix'),
                               fontsize=12*figscale,frameon=True,framealpha=1,loc='upper right',markerscale=1)
        else:
            legend = ax.legend((c1_label[0], c2_label[0], c3, c4),
                               ('Nullcline dA/dt = 0','Nullcline dB/dt = 0', 'Stable fixed point', 'Unstable fixed point'),
                               fontsize=12*figscale,frameon=True,framealpha=1,loc='upper right',markerscale=1)
            
    elif len(unstable_fp)==0:
        legend = ax.legend((c1_label[0], c2_label[0], c3),
                           ('Nullcline dA/dt = 0','Nullcline dB/dt = 0', 'Stable fixed point'),
                           fontsize=12*figscale,frameon=True,framealpha=1,loc='upper right',markerscale=1)
    
    contour_legends = legend.legendHandles[:2]
    for legend_obj in contour_legends:
        legend_obj.set_linewidth(3.0*figscale)
        
    return ax, sep
```

```python
def stability(x0, y0, p):
    # x0, y0 = fixed point coord
    # Calculate the stability of fixed point based on eigenvalue of jacobian matrix
    jmat = jacobian(x0, y0, p) # A Jacobian describes how each ODE rate changes when total A or B changes.
    eig = np.linalg.eigvals(jmat) # NumPy calculates the Jacobian's eigenvalues
    return (np.real(eig)<0).all() # True = locally stable; False = unstable. .all() requires both negative
```

The four long `jmat[...]` expressions are the four derivatives of the two ODEs with respect to the two **total** concentrations:

$$
J=\begin{pmatrix}
\dfrac{\partial (d[A_{\mathrm{tot}}]/dt)}{\partial [A_{\mathrm{tot}}]} &
\dfrac{\partial (d[A_{\mathrm{tot}}]/dt)}{\partial [B_{\mathrm{tot}}]} \\
\dfrac{\partial (d[B_{\mathrm{tot}}]/dt)}{\partial [A_{\mathrm{tot}}]} &
\dfrac{\partial (d[B_{\mathrm{tot}}]/dt)}{\partial [B_{\mathrm{tot}}]}
\end{pmatrix}.
$$

`jmat[0,0]` is the top-left entry; `jmat[0,1]` top-right; `jmat[1,0]` bottom-left; and `jmat[1,1]` bottom-right.

```python
def jacobian(x0, y0, p):
    # Calculate the jacobian matrix
    # Fixed point = (x0, y0)
    # p = parameters
    jmat = np.zeros([2,2])
    sqtemp = np.sqrt(p.Kd**2 + 8 * (x0 + y0) * p.Kd)
    
    jmat[0,0] = - 1 - p.beta * (1 + ((p.Kd + 4 * (x0 + y0) + sqtemp) / (2 * x0**2))**p.n)**(-2) * p.n * ((p.Kd + 4 * (x0 + y0) + sqtemp) / (2 * x0**2))**(p.n-1) * (x0 * (4 + 4 * p.Kd / sqtemp) - 2 * (p.Kd + 4 * (x0 + y0) + sqtemp)) / (2 * x0**3)
    jmat[0,1] = - p.beta * (1 + ((p.Kd + 4 * (x0 + y0) + sqtemp) / (2 * x0**2))**p.n)**(-2) * p.n * ((p.Kd + 4 * (x0 + y0) + sqtemp) / (2 * x0**2))**(p.n-1) * (4 + 4 * p.Kd / sqtemp) / (2 * x0**2)
    jmat[1,0] = - p.m * p.beta * (1 + (p.kappa * (p.Kd + 4 * (x0 + y0) + sqtemp) / (2 * y0**2))**p.n)**(-2) * p.n * (p.kappa * (p.Kd + 4 * (x0 + y0) + sqtemp) / (2 * y0**2))**(p.n-1) * p.kappa * (4 + 4 * p.Kd / sqtemp) / (2 * y0**2)
    jmat[1,1] = - p.gamma - p.m * p.beta * (1 + (p.kappa * (p.Kd + 4 * (x0 + y0) + sqtemp) / (2 * y0**2))**p.n)**(-2) * p.n * (p.kappa * (p.Kd + 4 * (x0 + y0) + sqtemp) / (2 * y0**2))**(p.n-1) * (y0 * p.kappa * (4 + 4 * p.Kd / sqtemp) - 2 * p.kappa * (p.Kd + 4 * (x0 + y0) + sqtemp)) / (2 * y0**3)
    
    return jmat
```

```python
def separatrix(x0, y0, df_dt, p, stepsize, Amax, Bmax):
    # Unstable fixed point = (x0, y0)
    # df_dt is ODE for A and B
    # Calculate separatrix
    # Quick check the first calculation goes to zero or inf.
    t = np.linspace(0,50,51)
    sep1 = scipy.integrate.odeint(df_dt, [x0, y0], -t, args=(p,))
    sep1 = sep1[~np.isnan(sep1).any(axis=1)]
    sep1 = sep1[~(sep1<0).any(axis=1)]
    if sep1[-1,0] > x0 and sep1[-1,1] > y0:
        if sep1[-1,0] < Amax:
            t = np.linspace(0,200,1001)
        else:
            t = np.linspace(0,50,2001)
        # Calculate the first half
        sep1 = scipy.integrate.odeint(df_dt, [x0, y0], -t, args=(p,))
        sep1 = sep1[~np.isnan(sep1).any(axis=1)]
        sep1 = sep1[~(sep1<0).any(axis=1)]
        # Calculate the second half
        t = np.linspace(0,200,1001)
        sep2 = scipy.integrate.odeint(df_dt, [x0-stepsize, y0-stepsize], -t, args=(p,))
        sep2 = sep2[~np.isnan(sep2).any(axis=1)]
        sep2 = sep2[~(sep2<0).any(axis=1)]
        sep = np.concatenate((np.flipud(sep1), sep2))
        sep = sep[sep[:,0]<1.1*Amax]
        sep = sep[sep[:,1]<1.1*Bmax]
        return sep[np.argsort(sep[:,0])]
    else:
        # Calculate the first half
        t = np.linspace(0,200,1001)
        sep1 = scipy.integrate.odeint(df_dt, [x0, y0], -t, args=(p,))
        sep1 = sep1[~np.isnan(sep1).any(axis=1)]
        sep1 = sep1[~(sep1<0).any(axis=1)]
        # Quick test second half
        t = np.linspace(0,50,51)
        sep2 = scipy.integrate.odeint(df_dt, [x0+stepsize, y0+stepsize], -t, args=(p,))
        sep2 = sep2[~np.isnan(sep2).any(axis=1)]
        sep2 = sep2[~(sep2<0).any(axis=1)]
        if sep2[-1,0] < Amax:
            t = np.linspace(0,50,2001)
        else:
            t = np.linspace(0,200,1001)
        # Calculate the second half
        sep2 = scipy.integrate.odeint(df_dt, [x0+stepsize, y0+stepsize], -t, args=(p,))
        sep2 = sep2[~np.isnan(sep2).any(axis=1)]
        sep2 = sep2[~(sep2<0).any(axis=1)]
        sep = np.concatenate((np.flipud(sep1), sep2))
        sep = sep[sep[:,0]<1.1*Amax]
        sep = sep[sep[:,1]<1.1*Bmax]
        return sep[np.argsort(sep[:,0])]
```

The `return` statement below evaluates the rates:

$$
\frac{d[A_{\mathrm{tot}}]}{dt}=\alpha+\beta\frac{[A_2]^n}{1+[A_2]^n}-[A_{\mathrm{tot}}],\qquad
\frac{d[B_{\mathrm{tot}}]}{dt}=r\alpha+m\beta\frac{[B_2]^n}{\kappa^n+[B_2]^n}-\gamma[B_{\mathrm{tot}}].
$$

```python
def df_dt(y, t, p):
    """
    ODE of A and B (considered asymmetry)
    """
    Atot, Btot = y # current total A and total B concentrations.
    A2 = 2*Atot**2 / (p.Kd + 4*(Atot+Btot) + np.sqrt(p.Kd**2 + 8 * (Atot+Btot) * p.Kd)) # calculate active [A2] directly from BOTH total concentrations.
    B2 = 2*Btot**2 / (p.Kd + 4*(Atot+Btot) + np.sqrt(p.Kd**2 + 8 * (Atot+Btot) * p.Kd)) # same for B2
    # `p` holds parameters such as alpha, beta, n, Kd, r, m, kappa, gamma.
    return [p.alpha + p.beta * A2**p.n / (1**p.n + A2**p.n) - Atot, 
            p.r * p.alpha + p.m * p.beta * B2**p.n / (p.kappa**p.n + B2**p.n) - p.gamma * Btot]
```

```python
class BSParams(object):
    """
    Container for parameters for homodimeric bistable circuits.
    """
    def __init__(self, **kwargs):
        # Dimensionless parameters
        self.alpha = 0.1*2 # leakiness
        self.beta = 10*2 # Maximal production
        self.Kd = 1 # Dissociation constant of dimers
        self.n = 1.5 # Additional cooperativity of transcriptional activation
        self.r = 1 # Leakiness ratio
        self.m = 1 # A and B copy number ratio
        self.kappa = 1 # Ratio of A and B's binding affinity with DNA
        self.gamma = 1 # Ratio of stability between A and B

        # Put in params that were specified in input
        for entry in kwargs:
            setattr(self, entry, kwargs[entry])
```

```python
class ThreeAffinityParams(BSParams):
    """Original author BSParams plus three independently adjustable affinities."""
    def __init__(self, KAA=1., KBB=1., KAB=1., **kwargs):
        super().__init__(**kwargs)
        self.KAA = float(KAA)
        self.KBB = float(KBB)
        self.KAB = float(KAB)
        if min(self.KAA, self.KBB, self.KAB) <= 0:
            raise ValueError('Dissociation constants must be strictly positive.')
```

For each guessed $[AB]$, the code uses the quadratic solution

$$
[A]=\frac{2([A_{\mathrm{tot}}]-[AB])}{1+\sqrt{1+8([A_{\mathrm{tot}}]-[AB])/K_{AA}}},\qquad
[B]=\frac{2([B_{\mathrm{tot}}]-[AB])}{1+\sqrt{1+8([B_{\mathrm{tot}}]-[AB])/K_{BB}}}.
$$

It keeps halving the allowed interval $0\leq[AB]\leq\min([A_{\mathrm{tot}}],[B_{\mathrm{tot}}])$ until the equilibrium condition is satisfied:

$$
[AB]-\frac{2[A][B]}{K_{AB}}=0.
$$

Then it returns $[A_2]=[A]^2/K_{AA}$ and $[B_2]=[B]^2/K_{BB}$. 

```python
def modified_dimers(Atot, Btot, p, steps=48):
    """Compute [A2] and [B2] from total concentrations, via [AB] bisection.

    Inputs may be scalars or NumPy grids (for plotting).
    Returns [A2], [B2], free [A], free [B], and equilibrium [AB].
    """
    Atot, Btot = np.broadcast_arrays(np.asarray(Atot, float),
                                     np.asarray(Btot, float))
    if np.any(Atot < 0) or np.any(Btot < 0):
        Atot = np.maximum(Atot, 0.)
        Btot = np.maximum(Btot, 0.)

    # From mass conservation, 0 <= [AB] <= min([Atot], [Btot]).
    # Start the allowed [AB] interval at zero for every concentration pair.
    lo = np.zeros_like(Atot)
    # [AB] cannot exceed either total concentration keep smaller upperbound
    hi = np.minimum(Atot, Btot)

    # Rearranged quadratic solution for each *free* monomer, at a guessed [AB].
    def free_monomer(amount_not_in_AB, K):
        # This is the stable quadratic formula.
        return 2*amount_not_in_AB / (1 + np.sqrt(1 + 8*amount_not_in_AB/K))

    # Root condition: [AB] - 2[A][B]/KAB = 0.
    # `range(steps)` repeats the search 48 times by default.
    for _ in range(steps):
        # Bisection: guess the midpoint of the currently possible [AB] interval.
        AB = (lo + hi) / 2
        # Using that guess, calculate the free A and B monomers.
        A = free_monomer(Atot - AB, p.KAA)
        B = free_monomer(Btot - AB, p.KBB)
        # Mass-action check: guessed [AB] minus the [AB] predicted by free monomers.
        # A positive error means the guess is too high and negative means too low.
        error = AB - 2*A*B/p.KAB
        # If guess is too low, move the lower bound up to the guessed [AB].
        lo = np.where(error < 0, AB, lo)
        # If guess is too high, move the upper bound down.
        hi = np.where(error >= 0, AB, hi)

    AB = (lo + hi) / 2
    A = free_monomer(Atot - AB, p.KAA)
    B = free_monomer(Btot - AB, p.KBB)
    # Convert the final free monomer values into the active homodimers:
    # [A2]=[A]^2/KAA; [B2]=[B]^2/KBB.
    A2 = A*A/p.KAA
    B2 = B*B/p.KBB
    # Return five concentrations, in the order given in the function docstring.
    return A2, B2, A, B, AB
```

Only the homodimer calculation changes. The production and degradation equations remain the authors' equations:

$$
\frac{d[A_{\mathrm{tot}}]}{dt}=\alpha+\beta\frac{([A]^2/K_{AA})^n}{1+([A]^2/K_{AA})^n}-[A_{\mathrm{tot}}],
$$

$$
\frac{d[B_{\mathrm{tot}}]}{dt}=r\alpha+m\beta\frac{([B]^2/K_{BB})^n}{\kappa^n+([B]^2/K_{BB})^n}-\gamma[B_{\mathrm{tot}}].
$$

```python
def modified_df_dt(y, t, p):
    """The authors' same ODE structure, evaluated with modified dimers."""
    Atot, Btot = y
    A2, B2, _, _, _ = modified_dimers(Atot, Btot, p)
    dA = p.alpha + p.beta*A2**p.n/(1 + A2**p.n) - Atot
    dB = p.r*p.alpha + p.m*p.beta*B2**p.n/(p.kappa**p.n + B2**p.n) - p.gamma*Btot
    return [dA, dB] # Return the instantaneous rates for both total concentrations.
```

The ODEs describe total concentrations, while the homodimer concentrations depend on free monomers. Therefore, we need to connect these two quantities before calculating the Jacobian.


$$
[A_{\mathrm{tot}}]=[A]+\frac{2[A]^2}{K_{AA}}+\frac{2[A][B]}{K_{AB}}
$$

$$
[B_{\mathrm{tot}}]=[B]+\frac{2[B]^2}{K_{BB}}+\frac{2[A][B]}{K_{AB}}
$$

Differentiate the conservation equations We want to determine how total concentrations change when free monomer concentrations change.

$$
[A_{\mathrm{tot}}]=[A]+\frac{2[A]^2}{K_{AA}}+\frac{2[A][B]}{K_{AB}}
$$

If we differentiate with respect to free A, treating free B as constant, we obtain:
$$
1+\frac{4[A]}{K_{AA}}+\frac{2[B]}{K_{AB}}
$$

The four derivatives are collected into a matrix:

$$
\boxed{
\begin{pmatrix}
1+\frac{4[A]}{K_{AA}}+\frac{2[B]}{K_{AB}} & \frac{2[A]}{K_{AB}}\
\frac{2[B]}{K_{AB}} & 1+\frac{4[B]}{K_{BB}}+\frac{2[A]}{K_{AB}}
\end{pmatrix}
}
$$

This is the mass_matrix: If free A or free B changes, how do the total concentrations change? 
For the Jacobian, we need the opposite relationship: If total A or total B changes slightly, how do the free monomer concentrations change?

We therefore calculate: np.linalg.inv(mass_matrix)  that calculates the inverse of a matrix.


Next, we need to determine how changing free A affects A production, and how changing free B affects B production.

Remember the Hill activation term for A:

$$
\beta\frac{[A_2]^n}{1+[A_2]^n}
$$

We know:

$$
[A_2]=\frac{[A]^2}{K_{AA}}
$$

Therefore, changing free A changes the amount of AA homodimer, which changes the activation-dependent production of A. **We calculate this relationship using the chain rule.**

First, differentiating the Hill production term with respect to the homodimer concentration $[A_2]$ gives:

$$
\frac{d}{d[A_2]}
\left(
\beta\frac{[A_2]^n}{1+[A_2]^n}
\right)
=
\beta\frac{n[A_2]^{n-1}}{(1+[A_2]^n)^2}
$$

Next, differentiating the homodimer concentration with respect to free A gives:

$$
\frac{d[A_2]}{d[A]}
=
\frac{2[A]}{K_{AA}}
$$

Using the chain rule, we multiply these two derivatives:

$$
\boxed{
\frac{d}{d[A]}
\left(
\beta\frac{[A_2]^n}{1+[A_2]^n}
\right)
=
\beta
\frac{n[A_2]^{n-1}}{(1+[A_2]^n)^2}
\frac{2[A]}{K_{AA}}
}
$$

Multiplying them gives the effect of changing free A on A production. We know how changes in total concentrations affect free monomer concentrations. We know how changes in free monomer concentrations affect transcriptional production. We combine these relationships using the **chain rule**. The resulting matrix tells us **how changes in total A and B affect the production rates of both transcription factors**.

However, our ODEs also include degradation. Therefore, we must also account for the degradation terms. Combining production and degradation gives the **complete Jacobian of our modified ODE system**. The Jacobian therefore describes how the rates of change of total A and B respond to small changes in their total concentrations.

Once we have identified a fixed point, we evaluate the Jacobian at that point and calculate its eigenvalues. **The eigenvalues tell us whether small disturbances around a fixed point grow or disappear over time.** If both eigenvalues have negative real parts, small disturbances decay, and the fixed point is **locally stable**

```python
def modified_jacobian(Atot, Btot, p):
    """Analytical Jacobian with the same free-monomer conservation equations."""
    A2, B2, A, B, AB = modified_dimers(float(Atot), float(Btot), p)
    # Derivatives of the two conservation equations with respect to free [A],[B].
    mass_matrix = np.array([
        [1 + 4*A/p.KAA + 2*B/p.KAB, 2*A/p.KAB],
        [2*B/p.KAB, 1 + 4*B/p.KBB + 2*A/p.KAB]
    ], dtype=float)

    # The chain rule: differentiation of the Hill production terms with respect
    # to free [A] and free [B]; the second line retains the authors' kappa.
    dprodA_dA = (p.beta*p.n*A2**(p.n-1) / (1+A2**p.n)**2) * (2*A/p.KAA)
    dprodB_dB = (p.m*p.beta*p.n*(p.kappa**p.n)*B2**(p.n-1)
                 / (p.kappa**p.n+B2**p.n)**2) * (2*B/p.KBB)

   # Chain rule: derivatives with respect to total A and total B.
   # `np.diag()` places the two production derivatives into a diagonal matrix.
  # `np.linalg.inv(mass_matrix)` calculates the inverse of the mass-conservation derivative matrix. This converts changes in total concentrations into changes in free monomers.
  # `@` performs matrix multiplication, combining these relationships through the chain rule.
    return (np.diag([dprodA_dA, dprodB_dB]) @ np.linalg.inv(mass_matrix)
            - np.diag([1., p.gamma]))


def modified_stability(Atot, Btot, p):
    """Same original stability criterion: both Jacobian eigenvalues negative."""
    return bool((np.real(np.linalg.eigvals(modified_jacobian(Atot, Btot, p))) < 0).all())  # Calculate the two eigenvalues and ask if BOTH real parts are negative.


print('Modified equilibrium functions are defined.')
```

## Validation (When the affinities are equal, do the models agree?)

First set $K_{AA}=K_{BB}=K_{AB}=K_d$. Also check $K_d=0.3,1,2$ to make sure the agreement is not limited to one value.

```python
validation_cases = [
    ('Mono high', 4., 100., 1., 1.5),
    ('Mono low', 0.1, 5., 1., 1.5),
    ('Quadrastability', 0.1, 12., 1., 1.5),
    ('Type I tri', 0.1, 8., 1., 1.5),
    ('Type II tri (15)', 0.2, 15., 1., 1.),
    ('Type II tri (20)', 0.8, 20., 1., 1.5),
    ('Bistability', 0.4, 10., 1., 1.5),
    ('Legends (quad)', 0.1, 17.5, 10., 2.),
]

for kd in [0.3, 1., 2.]:
    author_p = BSParams(Kd=kd, alpha=.8, beta=20., n=1.5)
    mod_p = ThreeAffinityParams(KAA=kd, KBB=kd, KAB=kd,
                                Kd=kd, alpha=.8, beta=20., n=1.5)
    probe = np.array([0.05, .2, 1., 3., 6., 12., 25.])
    X, Y = np.meshgrid(probe, probe)

    author_A2 = 2*X**2/(kd+4*(X+Y)+np.sqrt(kd**2+8*kd*(X+Y)))
    author_B2 = 2*Y**2/(kd+4*(X+Y)+np.sqrt(kd**2+8*kd*(X+Y)))
    A2, B2, _, _, _ = modified_dimers(X, Y, mod_p)
    assert np.allclose([author_A2, author_B2], [A2, B2], atol=1e-10, rtol=1e-10) # `np.allclose` checks numerical agreement despite tiny rounding errors.

    original_rates = np.stack(df_dt([X, Y], 0., author_p))
    modified_rates = np.stack(modified_df_dt([X, Y], 0., mod_p))
    assert np.allclose(original_rates, modified_rates, atol=1e-10, rtol=1e-10)

    for Atot, Btot in [(0.6, 0.8), (3., 8.), (6., 12.), (15., 20.)]:
        assert np.allclose(jacobian(Atot,Btot,author_p),
                           modified_jacobian(Atot,Btot,mod_p), atol=1e-7,rtol=1e-7)
    print(f'PASS: all dimer, ODE, and Jacobian comparisons for K={kd:g}')

print('PASS: all three equal-affinity tests agree with the AUTHORS\' source functions.')
```

## Find fixed points for both models

A **fixed point** is a concentration pair where both ODE rates are zero. We search from many initial guesses because there can be several fixed points. Then we evaluate each model's Jacobian to classify each fixed point as stable or unstable.

```python
from scipy.optimize import root
from scipy.optimize import linear_sum_assignment
from matplotlib.colors import ListedColormap
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import seaborn as sns

REGIMES = [
    dict(name='Monostability (high)', alpha=4., beta=100., Kd=1., n=1.5, rangescale=2., tickint=50),
    dict(name='Monostability (low)', alpha=0.1, beta=5., Kd=1., n=1.5, rangescale=.8, tickint=2),
    dict(name='Quadrastability', alpha=0.1, beta=12., Kd=1., n=1.5, rangescale=1.2, tickint=4),
    dict(name='Type I tristability', alpha=0.1, beta=8., Kd=1., n=1.5, rangescale=1.2, tickint=4),
    dict(name='Type II tristability (15)', alpha=0.2, beta=15., Kd=1., n=1., rangescale=1.5, tickint=10),
    dict(name='Type II tristability (20)', alpha=0.8, beta=20., Kd=1., n=1.5, rangescale=2.5, tickint=10),
    dict(name='Bistability', alpha=0.4, beta=10., Kd=1., n=1.5, rangescale=1.5, tickint=4)
]

ORIGINAL_STATE_COLORS = {(1,1):0, (1,0):1, (0,1):2, (0,0):6}




def make_params(regime, modified=False, KAA=None, KBB=None, KAB=None):
    """Set each dimerization constant INDEPENDENTLY for the modified model.

    When a K is omitted, it defaults to the paper's Kd for this regime.
    Example: make_params(regime, modified=True, KAA=0.3, KBB=1, KAB=2)
    """
    common = dict(alpha=regime['alpha'], beta=regime['beta'],
                  n=regime['n'], Kd=regime['Kd'], r=1., m=1.,
                  kappa=1., gamma=1.)
    if not modified:
        return BSParams(**common)  # Original author's one-Kd parameters.
    return ThreeAffinityParams(
        KAA=regime['Kd'] if KAA is None else KAA,
        KBB=regime['Kd'] if KBB is None else KBB,
        KAB=regime['Kd'] if KAB is None else KAB,
        **common)
```

A **smaller dissociation constant means stronger dimerization**. 

For a fair comparison across the authors’ regimes (including their $K_d=10$ example), the three numbers are initially entered as **multiples of that regime’s original $K_d$**. When $K_d=1$, the numbers are the actual $K$ values.

- **AA favored:** $K_{AA}=0.3K_d$, $K_{BB}=K_{AB}=K_d$. Tests an advantage for *one* homodimer.
- **Both homodimers favored:** $K_{AA}=K_{BB}=0.3K_d$, $K_{AB}=K_d$. Tests whether stronger self-activation relative to heterodimer sequestration changes multistability.
- **Heterodimer favored:** $K_{AA}=K_{BB}=K_d$, $K_{AB}=0.3K_d$. Tests stronger mutual sequestration of A and B.

```python
AFFINITY_CASES = [
    dict(name='AA favored',           KAA=0.3, KBB=1.0, KAB=1.0),
    dict(name='Both homodimers favored', KAA=0.3, KBB=0.3, KAB=1.0),
    dict(name='Heterodimer favored',   KAA=1.0, KBB=1.0, KAB=0.3),
]

# True = the numbers above are multiplied by the current regime's original Kd.
# False = the numbers above ARE the absolute, dimensionless K values.
RELATIVE_TO_ORIGINAL_KD = True

# One-line summary of the scenarios, as they will apply to a Kd=1 regime:
for case in AFFINITY_CASES:
    print(f"{case['name']:<26} KAA={case['KAA']}, KBB={case['KBB']}, KAB={case['KAB']}")
```

A fixed point is a concentration pair where **both** ODE rates vanish. `scipy.optimize.root` tries starting concentrations until both residuals are close to zero.  This root search is **different from the bisection step**, which computes the monomer/dimer equilibrium for given totals.

```python
def fixed_points(p, modified=False, nstarts=14):
    """Find fixed points, retaining the authors' stability criterion.

    For modified equilibria, solve directly for the free [A],[B] concentrations
    (no re-solving the bisection loop inside every root search).
    """
    upper = max(float(p.alpha+p.beta), float(p.r*p.alpha+p.m*p.beta))
    if modified:
        # For the modified model, search using FREE [A],[B] instead of total A,B.
        def residual_ab(ab):
            A,B = ab
            # Reject unphysical guesses with negative free-monomer concentrations.
            if A < 0 or B < 0:
                # Smooth continuation is preferable to clipping during root search.
                return np.array([A - abs(A) - 1, B - abs(B) - 1]) * 1e2
            A2, B2 = A*A/p.KAA, B*B/p.KBB # Homodimers from the modified mass-action equations.
            AB = 2*A*B/p.KAB # Inactive heterodimer from its mass-action equilibrium.
            Atot = A + 2*A2 + AB # Use conservation of mass to recover total A and total B.
            Btot = B + 2*B2 + AB
            return np.array([
                p.alpha + p.beta*A2**p.n/(1+A2**p.n) - Atot,
                p.r*p.alpha + p.m*p.beta*B2**p.n/(p.kappa**p.n+B2**p.n) - p.gamma*Btot
            ]) # Two residuals = production minus degradation for A and B A fixed point is found when BOTH residuals are zero.
        guesses = np.geomspace(.02, max(1.0, np.sqrt(upper*max(p.KAA,p.KBB))), nstarts) # Multiple starts help discover more than one fixed point.
        roots=[]
        for A in guesses:
            for B in guesses:
                r = root(residual_ab, [A,B], method='hybr', options={'maxfev':1000})
                if not r.success or np.linalg.norm(residual_ab(r.x))>1e-5:
                    continue
                af,bf=r.x
                if min(af,bf)<0:continue
                a2,b2=af**2/p.KAA,bf**2/p.KBB
                at=af+2*a2+2*af*bf/p.KAB
                bt=bf+2*b2+2*af*bf/p.KAB
                if not any(np.linalg.norm(np.array([at,bt])-np.array(existing[:2]))<1e-3 for existing in roots):
                    roots.append([at,bt,modified_stability(at,bt,p), a2,b2])
    else:  # For the original model, solve the author ODEs directly in total A and B.
        def fun(z):
            a,b=z
            if a<0 or b<0:return np.array([a-abs(a)-1,b-abs(b)-1])*1e2
            return np.asarray(df_dt([a,b],0.,p),float)
        guesses=np.geomspace(.02, max(1.,upper),nstarts)
        roots=[]
        for a in guesses:
            for b in guesses:
                r=root(fun,[a,b],method='hybr', options={'maxfev':1000})
                if not r.success or np.linalg.norm(fun(r.x))>1e-5 or np.min(r.x)<0:
                    continue
                at,bt=r.x
                if not any(np.linalg.norm(np.array([at,bt])-np.array(existing[:2]))<1e-3 for existing in roots):
                    d= p.Kd+4*(at+bt)+np.sqrt(p.Kd*p.Kd+8*p.Kd*(at+bt))
                    roots.append([at,bt,bool(stability(at,bt,p)),2*at*at/d,2*bt*bt/d])
    roots.sort(key=lambda row:(-row[2],row[0],row[1]))
    return roots


def check_equal_fixed_points(regime):
    # Build matched original and equal-affinity modified model parameters.
    po=make_params(regime)
    pm=make_params(regime, modified=True)
    # Find fixed points independently in each model.
    ro=fixed_points(po,False)
    rm=fixed_points(pm,True)
    assert len(ro)==len(rm),f"{regime['name']}: root count differs: {len(ro)}, {len(rm)}"
    pos_o=np.asarray([r[:2] for r in ro]);pos_m=np.asarray([r[:2] for r in rm])
    costs=np.linalg.norm(pos_o[:,None,:]-pos_m[None,:,:],axis=-1)
    ii,jj=linear_sum_assignment(costs)
    assert np.max(costs[ii,jj])<1e-4,(regime['name'], costs[ii,jj])
    for i,j in zip(ii,jj):
        assert ro[i][2]==rm[j][2], f"{regime['name']}: stability differs"
    return len(ro), sum(r[2] for r in ro),np.max(costs[ii,jj])

print('Fixed-point checks across all original plotting regimes:')
for r in REGIMES:
    total, stable, err=check_equal_fixed_points(r)
    print(f"PASS {r['name']:<29} fixed points={total}; stable={stable}; max location difference={err:.2e}")
```

## Ploting Functions

```python
def basin_labels(p, modified, roots, Amax, Bmax, density=72, steps=105, dt=.3):
    """Integrate a grid of starting conditions to identify attraction basins.

    Basin shading is an explicitly new numerical routine; the gray flow lines,
    orange/red nullclines, and black/white dots are drawn by the  author `plot_flow_field` function.
    """
    stable = [r for r in roots if r[2]]
    if not stable:raise ValueError('No stable fixed point was found for this regime.')
    u = np.linspace(0., Amax, density)
    v = np.linspace(0., Bmax, density)
    Atot,Btot=np.meshgrid(u,v)
    ode = modified_df_dt if modified else df_dt
    # Euler integration with dt=0.3 on the positive state space; continue long
    # enough for almost all points to approach an attracting fixed point.
    for _ in range(steps):
        dAtot,dBtot = ode([Atot,Btot],0.,p)
        Atot = np.clip(Atot + dt*dAtot,0.,Amax*2)
        Btot = np.clip(Btot + dt*dBtot,0.,Bmax*2)

    target=np.asarray([r[:2] for r in stable])
    nearest=np.argmin((Atot[...,None]-target[:,0])**2+(Btot[...,None]-target[:,1])**2,axis=-1)
    # Assign the authors' original palette index to each attracting expression state.
    indices=[]
    for fp in stable:
        expression=(int(fp[3]>1.),int(fp[4]>1.))
        indices.append(ORIGINAL_STATE_COLORS[expression])
    categories=np.asarray(indices)[nearest]
    return u,v,categories,stable
```

```python
def display_author_style_panel(ax, p, modified, extent, title, tickint=5, grid_density=185):
    """Use exact author flow-field function and same author color palette."""
    Amax,Bmax=extent
    roots=fixed_points(p,modified)
    u,v,category,stable=basin_labels(p,modified,roots,Amax,Bmax)
    # Put translucent category colors behind the author's line drawing.
    pal=sns.color_palette()
    # Values 0,1,2,6 are the same author's palette indices as in their plots.
    color_table=[pal[0],pal[1],pal[2],pal[6]]
    remap=np.zeros_like(category)
    for label,ind in enumerate([0,1,2,6]):
        remap[category==ind]=label
    ax.imshow(remap,origin='lower',extent=(0,Amax,0,Bmax),
              interpolation='nearest',cmap=ListedColormap(color_table),
              alpha=.25,zorder=0,aspect='auto',vmin=-.5,vmax=3.5)

    # Exact plot_flow_field source from uploaded original author notebook.
    # Its `stability` name is looked up when the function executes, so for the
    # modified panel ONLY we temporarily select the new Jacobian criterion.
    original_stability_function=globals()['stability']
    if modified:
        globals()['stability']=modified_stability
    try:
        with warnings.catch_warnings():
            warnings.filterwarnings('ignore',message='No contour levels were found')
            ax,_ = plot_flow_field(ax, modified_df_dt if modified else df_dt,
                                   [max(1e-5, p.alpha*.01),Amax],
                                   [max(1e-5, p.r*p.alpha*.01),Bmax],
                                   p,density=grid_density,sep_fun=None,figscale=.38)
    finally:
        globals()['stability']=original_stability_function

    # Same author's dashed green line for a separatrix. Here contours of
    # forward-estimated basin categories replace hand-built author polygons.
    if len(set(category.ravel().tolist()))>1:
        # One boundary for each basin category; avoids assuming how many basins
        # exist under unequal affinities.
        for label in sorted(set(category.ravel().tolist()))[:-1]:
            binary=(category==label).astype(float)
            ax.contour(u,v,binary,levels=[.5],colors=[pal[2]],linestyles='--',
                       linewidths=1.6,zorder=3,alpha=.9)

    # The authors' unchanged formatting function calls plt.show() and
    # plt.tight_layout() for standalone figures. Suppress only those two calls
    # while composing our four panels on one figure.
    show_original, layout_original = plt.show, plt.tight_layout
    plt.show = lambda *args, **kwargs: None
    plt.tight_layout = lambda *args, **kwargs: None
    try:
        plotphaseportrait(
            ax, [p.alpha, p.beta, p.Kd, p.n, p.r, p.m, p.kappa, p.gamma],
            [p.alpha*.01, Amax], [p.r*p.alpha*.01, Bmax],
            figscale=.38, tickint=tickint, savedir=None,
            display_legend=False, basins=None)
    finally:
        plt.show, plt.tight_layout = show_original, layout_original
    ax.set_title(title, fontsize=11, pad=10)
    ax.set_aspect('equal', adjustable='box')
    # Draw fixed point markers again on top of the separatrix/basin contours.
    for fp in roots:
        ax.scatter(fp[0],fp[1],s=68,facecolor='black' if fp[2] else 'white',
                   edgecolor='black',linewidth=1.7,zorder=8)
    return roots


def compare_regime(regime, save=True):
    """Plot original, equal-affinity control, then THREE independent affinity cases."""
    kd = regime['Kd']
    original_p = make_params(regime)
    equal_p = make_params(regime, modified=True, KAA=kd, KBB=kd, KAB=kd)
    panels = [
        (f"Original paper\n$K_d={kd:g}$ (all dimers)", original_p, False),
        (f"Equal-K control\n" + rf"$K_{{AA}}={kd:g},\,K_{{BB}}={kd:g},\,K_{{AB}}={kd:g}$", equal_p, True),
    ]

    for case in AFFINITY_CASES:
        scale = kd if RELATIVE_TO_ORIGINAL_KD else 1.0
        kaa = case['KAA'] * scale
        kbb = case['KBB'] * scale
        kab = case['KAB'] * scale
        model_p = make_params(regime, modified=True, KAA=kaa, KBB=kbb, KAB=kab)
        label = (f"{case['name']}\n"
                 + rf"$K_{{AA}}={kaa:g},\,K_{{BB}}={kbb:g},\,K_{{AB}}={kab:g}$")
        panels.append((label, model_p, True))

    Amax = Bmax = regime['rangescale'] * regime['beta']
    fig, axs = plt.subplots(1, len(panels), figsize=(5.15*len(panels), 5.4),
                            sharex=True, sharey=True)
    results = []
    for ax, (label, p, is_modified) in zip(axs, panels):
        points = display_author_style_panel(ax, p, is_modified, (Amax, Bmax),
                                             label, tickint=regime['tickint'])
        results.append((len(points), sum(point[2] for point in points)))
    fig.suptitle(f"{regime['name']}:  "
                 + rf"$\alpha={regime['alpha']}$, $\beta={regime['beta']}$, "
                 + rf"$n={regime['n']}$, original $K_d={kd:g}$",
                 y=1.055, fontsize=16)
    fig.subplots_adjust(wspace=.14, top=.88)
    if save:
        name = (regime['name'].replace(' ', '_').replace('(', '').replace(')', '')
                .replace('—', '_').replace('/', '_') + '_three_biology_cases.png')
        filepath = 'MF2_' + name
        fig.savefig(filepath, dpi=125, bbox_inches='tight')
    else:
        filepath = None
    plt.close(fig)
    from IPython.display import Image, display
    if filepath:
        display(Image(filename=filepath))
    label_names = ['Original', 'Equal-K'] + [case['name'] for case in AFFINITY_CASES]
    print('Fixed points / stable states (per panel):')
    for label, numbers in zip(label_names, results):
        print(f"  {label:<26} {numbers[0]} fixed, {numbers[1]} stable")
    return filepath, results
```
